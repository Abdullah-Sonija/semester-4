### 3.7 Exploring Data 

```python
numeric_features = [feature for feature in df.columns if df[feature].dtype != 'O']
categorical_features = [feature for feature in df.columns if df[feature].dtype == 'O']

print('We have {} numerical features : {}'.format(len(numeric_features), numeric_features))
```

```python
df.columns
```

```python
df['Cpu'] = df['Cpu'].fillna('Unknown')
```

```python
def extract_cpu_brand(text): 
    # Convert input to string to prevent float errors
    text = str(text)
    
    if 'Intel' in text: 
        return 'Intel' 
    elif 'AMD' in text: 
        return 'AMD' 
    elif 'Apple' in text: 
        return 'Apple' 
    elif 'Qualcomm' in text: 
        return 'Qualcomm' 
    else: 
        return 'Other' 
 
df['Cpu_Brand'] = df['Cpu'].apply(extract_cpu_brand) 
df['Cpu_Brand'].value_counts()
```

```python
categorical_features.append('Cpu_Brand') 
```
